```python
import numpy as np
import time
import pandas as pd
import pymysql
data=pd.read_csv(r'loan_data\converted_training_data.csv')
test_data=pd.read_csv(r'loan_data\converted_test_data.csv')
df=pd.concat([data,test_data])
df.LoanAmount=df.LoanAmount.apply(lambda x:x+10000)
X=df.drop(['LoanAmount','Property_Area','Education'],axis=1)
y=df.LoanAmount
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X,y)

le_d={'MALE':1,'FEMALE':0,'OTHERS':2,'YES':1,'NO':0,'0':0,'1':1,
     '2':2,'3':3,'3+':3,'SELF':2,'ORGANIZATION':0,'UNEMPLOYED':1}

def data_pre(n,g,m,d,s,ai,ei,la,lat,llp):
    gen=le_d[g]
    mar=le_d[m]
    dep=le_d[d]
    sel=le_d[s]
    last=le_d[llp]
    return n,la,np.array([gen,mar,dep,sel,ai,ei,lat,last]).reshape(1,-1)

    
def user_data():
    name=input('Enter your name ').upper()
    #name='RAJA'
    mail=input('Enter your mail id')
    
    gen=input('Enter your gender(male/female/others) ').upper()
    #gen='OTHERS'
    
    mar=input('Are you Married?(yes/no) ').upper()
    #mar='NO'
    
    if(mar=='YES'):
        dep=input('how many childs you have ?(0/1/2/3+) ')
    else:
        dep='0'
    #dep='0'
    
    emp=input('Employment Status (self/organization/unemployed) ').upper()
    #emp='SELF'
    
    apin=int(input('enter your yearly income '))
    #apin=500000
    
    exin=int(input('enter your Extra income '))
    #exin=3000
    
    loan_amt=int(input('how much amount of loan do you want? '))
    #loan_amt=20000
    
    loan_amt_t=int(input('enter loan ammount period in days (min 12)'))
    #loan_amt_t=45
    
    
    l_l_p=input('Did you pay your last loan?(yes/no) ').upper()
    #l_l_p='YES'
    
    usr_data=data_pre(name,gen,mar,dep,emp,apin,exin,loan_amt,loan_amt_t,l_l_p)

    pre=model.predict(usr_data[2])
    if(pre[0]<usr_data[1]):
        print(f'Hello {usr_data[0]} you are not elible to get {usr_data[1]} amount loan')
        time.sleep(5)
        print(f'But you can take {int(pre[0])} amount as a loan!!!')
    else:
        print(f'Congratulations {usr_data[0]} you can take {usr_data[1]} amount loan!')
    
    data=(name,gen,mail,apin,int(pre[0]))
    print('con...making...')
    db=pymysql.connect(user='root',password='',host='localhost',database='BBD')
    mycursor=db.cursor()
    print('connection created')
    q='''insert into Loan_DATA values(%s,%s,%s,%s,%s)''' 
    mycursor.execute(q,data)
    print("query executed")
    db.commit()


           
user_data()     
```
